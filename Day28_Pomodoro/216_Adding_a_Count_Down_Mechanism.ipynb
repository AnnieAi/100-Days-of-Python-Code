# 216.1 TIMER MECHANISM

````python
def start_timer():
    count_down(5 * 60)
````

count_down是需要传参的，但是加给button的command的只是函数对象，不能传参，所以要多修一层装饰、入口？用这个方法来实现按下按钮就能开始countdown。

# 216.2 COUNTDOWN MECHANISM

````python
def count_down(count):

    count_min = math.floor(count / 60)
    count_sec = count % 60
    
    canvas.itemconfig(timer_text, text=f"{count_min}:{count_sec}")
    
    if count > 0:   # count = 0时，停止递归（使用递归时需要设置终止语句）
        window.after(1000, count_down, count - 1)     # 1000 毫秒后，调用 count_down(count - 1)
````

现阶段的问题在于，5分钟整的时候，显示的是5:0而不是5:00。这个问题要靠Python dynamic coding解决。这是下一课的内容。

### item ID

```python
canvas.itemconfig(timer_text, text=count)
```
这里只是为了改label。如果是Label类的对象，直接写：label_object.config(text="") 就行了。但这里的timer_text是canvas上的文字对象，而不是一个单纯的Label类对象，所以要用itemconfig方法。

### `after`方法

#### 方法来源：
tkinter Module - Misc Class - after Method

#### 方法声明：
`window.after(ms, func, *args)`

1. `ms`: 等待时间，单位是毫秒
2. `func`: 等待结束后要执行的函数，不加括号
3. `*args`: 传给 func 的参数

#### args的专门说明

1. args到底对应多少个参数？

   不是只能传一个参数，它是“后面所有要传给函数的参数”。也就是传进去的参数数目和这个函数能够承接的参数数目要对应。
  
    比如：`window.after(1000, count_down, count - 1, 4, 8)`  传进的参数数目就是3个，而不是要求的1个。这时候就会报错。


2. 具体在哪里触发报错？

   按照上面这样写，`window.after()`本身是不会报错的，因为它实际上只是一个...标记。

   真正报错的时间点是在下一次调用函数`count_down(count)`，尝试一口气塞3个参数 `count-1, 4, 8` 进 `count_down`的 `count` 时

3. 报错信息
    > func(*args)
    >
    > TypeError: count_down() takes 1 positional argument but 3 were given


# 215. UI SETUP

## Window

````python
# Window
window = Tk()
window.title("Pomodoro")
window.config(padx=100, pady=50, bg=YELLOW)

## Label1
timer_label = Label(text="Timer", font=(FONT_NAME, 50), fg=GREEN, bg=YELLOW)
timer_label.grid(column=1, row=0)

## Canvas
canvas = Canvas(width=200, height=224, bg=YELLOW, highlightthickness=0)
tomato_img = PhotoImage(file="tomato.png")
canvas.create_image(100, 112, image=tomato_img)
timer_text = canvas.create_text(100, 130, text="00:00", fill="white", font=(FONT_NAME, 35, "bold"))
canvas.grid(column=1, row=1)
````

````python
## Button1
start_button = Button(text="Start", highlightthickness=0, command=start_timer)
start_button.grid(column=0, row=2)
````

这里做了修改。目的是为了点击start按钮的时候就会开始倒计时。

````python
## Button2
reset_button = Button(text="Reset", highlightthickness=0)
reset_button.grid(column=2, row=2)

# Label2
check_marks = Label(text="✔", fg=GREEN, bg=YELLOW)
check_marks.grid(column=1, row=3)

window.mainloop()
````
