```python
from tkinter import *

window = Tk()
window.title("My First GUI Program")
window.minsize(width=500, height=300)
```

## Label

```python
my_label = Label(text="I Am a Label", font=("Arial", 24, "bold"))
my_label.pack()
```

```python
### Tkinter 有两种改控件属性的方法，它们在这里效果一样。

my_label["text"] = "New Text"           # Tkinter 控件对象自己提供的“字典风格配置写法”
my_label.config(text="New Text")        # 调用config方法
```

`my_label["text"] = "New Text"`  这是 Tkinter 控件对象自己提供的“字典风格配置写法”。

而想要一次改好几个属性，就用config。

## Button

```python
def button_clicked():
    print("I got clicked!")


button = Button(text="Click me", command=button_clicked)
button.pack()
```

command起到的作用类似于listener，在检测到button被click后，就会调用command后面的这个函数。

注意，传给command的是函数的名字，而不是函数，所以是`button_clicked`而不是`button_clicked()`，不要带括号。

```python
# Challenge 1  点击按钮后修改label显示的文本

def change_text():
    my_label.config(text="Button Got Clicked")

button1 = Button(text="CHALLENGE1", command=change_text)

button1.pack(side="left")
```

## Entry

```python
input = Entry(width=10)
input.pack()
```

```python
# Challenge 2  想办法让用户输入的内容能够显示在label上

def show_input():
    input_string = input.get()
    my_label.config(text=input_string)

button2 = Button(text="CHALLENGE2", command=show_input)

button2.pack(side="right")
```
