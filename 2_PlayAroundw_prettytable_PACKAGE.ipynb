## 1. About prettytable

The `prettytable` we installed using prior bash commands is a PACKAGE.

And we are going to fetch the <class> `PrettyTable` from the package.

````python
from prettytable import PrettyTable    # from prettytable PACKAGE get PrettyTable CLASS
````

Then we use create a `PrettyTable` class object. 

````python
table = PrettyTable()
print(table)    # This line of code prints the backbone of a Pretty table, it is difficult to decipher it though
````

We can see that although we haven't made any configuration yet, the object `table` already has the backbone of a PrettyTable object.

(Note that the output in Jupyternotebook is not like the original ASCII it should have in terminals, if you want to view the original ASCII output, please refer to the PrettyTableTask.py)

## 2. Use Method

````python
table.add_column("Pokemon Name",["Pikachu", "Squirtle", "Charmander"])
table.add_column("Type",["Electric", "Water", "Fire"])

print(table)
````

For `PrettyTable` <class> objects, we use method:

`object.add_column(fieldname,column)`

To create a column. Where `fieldname` should be a **string** that is the title of this column, and the `column` is a **list** that contains elements you want to put into the column (can be strings, integers, etc.).

## 3. Set Attribute
We can set the `PrettyTable` <class> object `table`'s attribute `align` to `l`, which means left-aligned. Before, it is center-aligned (still unable to view in Jupyternotebook, please refer to the original .py file), now it becomes left-aligned.

````python
table.align = "l"
print(table)
````

## 4. More to Learn
Maybe you've noticed that when dealing with `turtle`, we say it's a *module*, but when dealing with `prettytable`, we say it's a *package*. So what do these terminologies mean?

Well, in the context of Angela Yu's "100 Days of Code" (and most Python conversations), the distinction between module and package is usually drawn like this:

1. **`turtle` is a Module:**
* It comes **built-in** with Python (the Standard Library).
* It is just a file (like `turtle.py`) sitting in your Python installation.
* You don't need to install anything; you just `import` it.


2. **`prettytable` is a Package:**
* It is **external** (Third-Party).
* It is a collection of code that you had to go out and "buy" (install) from the Python Package Index (PyPI).
* We generally call anything we install via `pip` a "package."



### The "Super Technical" Nuance (Just for your info)

Technically, in Python:

* A **Module** is a single file (e.g., `code.py`).
* A **Package** is a **folder** containing multiple modules and a special `__init__.py` file.

*Fun Fact:* `prettytable` (especially the older version 0.7.2 you are using) is actually just a single file (`prettytable.py`) inside your computer, so technically it is *also* a module! But because we download it as a "distributable unit," everyone calls it a **package**.

**TL;DR:** Stick with your definition.

* **Module** = Built-in tool (Standard Library).
* **Package** = Downloadable addon (PyPI).
