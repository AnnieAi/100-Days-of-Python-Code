````python
import json
````

# 1 创建一个空的json文件

这行代码就是创建 JSON 文件的标准做法：

```python
with open("data.json", "w") as file:
```

其中，`"w"` 是 write mode：
- 文件不存在：创建新文件
- 文件存在：清空旧内容，再写入新内容

## 1.1 创建一个完全空的JSON文件

它是全空的文件。本质上并不是一个有效的 JSON 数据文件。

````python
with open("data.json", "w") as file:
    pass
````

## 1.2 创建一个里面有`{}`的 JSON 文件

这个是更推荐的写法。有了`{}`就是有效的JSON文件了。

````python
with open("data.json", "w") as file:        
    json.dump({}, file, indent=4)
````

# 2 往已经存在的 JSON 文件中写入更多的数据

````python
students = {
    "Alice": {
        "math": 90,
        "english": 85
    },
    "Bob": {
        "math": 75,
        "english": 88
    }
}
````

````python
with open("data.json", "w") as file:     
    json.dump(students, file, indent=4)
````

```python
json.dump(students, file, indent=4)
```
把 Python 字典 students 转成 JSON 格式，写进文件。
