## 1. Intro

从一个class，可以构建很多object（对象），我们会在代码里真正使用和操作的正是object。

Normally a class is written with the first letter of **each** word capitalized, which is known as Pascal case (帕斯卡命名法).
 
> PascalCase: The first letter of every word is capitalized.
>
> camelCase: Every word's first letter is capitalized desipte the first word's.
>
> snake_case: All words are in lower case and connected by _.
>
> In Python, camelCase doesn't appear frequently. We use **PascalCase for class name** and **snake_case for all others**.

This is to differentiate it from all the variable and function names that we give in Python, where each word is separated by underscores.

```python
car = CarBlueprint()    # "Car" and "Blueprint"
```

In this case, the `car` is the **object** and it gets created from this `CarBlueprint()`.

All we have to do to create the object from the class is to give the object a name, it can be anything you want, like the `car` here, set it equal to the name of the class, like the `CarBlueprint`, and the parentheses, which in the same way as it activates the function, it activates **the construction** of this object from the class.

## 3. 使用Turtle Module

### 3.1 Method 1

Now we wanna tab into the class called turtle that's declared in this turtle. The wording is a little bit confusing, let me explain that:

````python
# <originally from 1_CreateMyOwnModule_Use_turtle_MODULE.py>

import turtle               # 导入turtle模块/turtle module

timmy = turtle.Turtle()     # 使用turtle模块的Turtle类的构造函数Turtle()来construct一个Turtle类的object
````

1. The first "turtle": Tab into the `turtle` *module*.
2. The second "Turtle": Get hold of the `Turtle` *class*.
3. The parentheses: To actually construct the object, we need to add the parentheses at the end.
4. The "Timmy": Save all of these into an object.

So actually until now we've done something as we done with the `another_variable` module.
1. We both use `import` to import the module.
2. We tab into the module.
3. We fetch something from the module. In `another_variable`, it's a variable; in `turtle`, it's a class.

## 4. Object Attribute

Now, as we saw before with our car object, it has certain attributes, like speed and fuel, data that it could keep track of that are really important to modeling of an actual car object.
Now when it comes to accessing these attributes, the syntax or the code looks like this:
```python
car.speed
```
Car in this case is the object.
We use a dot or a full stop to separate the object from the actual attribute that we want to get hold of, which in this case is the speed.
What this code does is it essentially identifies the object and then it says, from this object get the speed attribute. This is actually gonna represent the speed of this particular car object.



Let's go back to our `turtle` module.
One of the other class that is inside this turtle module is called `Screen`.

````python
from turtle import Turtle, Screen
````

And the `Screen` represents the window in which this turtle is going to show up.

So let's go ahead and create this screen object.

````python
my_screen = Screen()
````

Actually when we are running the actual code, when the program finishes conducting this line of code, a window pops up for an instant. THIS line of code CREATES the window.

Now we can tap into one of the screen's properties.

````python
print(my_screen.canvheight)
````

So the `my_screen` here is the object and that `canvheight` is an attribute that's associated with that object.

## 5. Object Methods

Methods are FUNCTIONS that are tied to an object. But essentially, we would use it the same way as we have done with any other function that we've created so far.

The only different is the syntax.

First, we would tap into the object and using that same dot notation that we saw with the attributes
```python
car.stop()
```
So essentially what we're doing is we're getting hold of the object and then we're calling this function that's associated with that object, which is known as a method.

````python
my_screen.exitonclick()
````

Now let's tap into one of the `my_screen`'s method `exitonclick`.

And what this will allow us to do is instead of having our screen show up and then quickly disappear when our code ends, this `exitonclick` will allow our program to continue running until we click on the screen and it exits our code.

Besides these, we can also change the shape and color of the turtle. Referring to the `main.py` to see actual codings.
