# 1. Catching Exceptions 的总框架：4个keyword组成的结构

In programming, we can catch these exceptions.

When sth goes wrong and at that moment we catch that exception,then it doesn't have to fail catastrophically.

We can actually fail more gracefully or we can decide that sth else should happen.

Below are the keywords that are used in catching exceptions:

```python
try:

except:

else:

finally
```

## 1.1 View these keywords in detail

```python
try:
```
You are executing sth that might cause an exception.

```python
except:
```
The block of code you want the computer to execute if there was an exception.

```python
else:
```
The block of code you want the computer to execute if there was no exception.

```python
finally:
```
This block of code is executed no matter what happens.

## 1.2 In what sequence they are actually executed?

```python
try:

except:

else:

finally
```

这是最完整的表达结构。其中，`except` 可以有多个。

这个结构也可以去除掉一部分使用，接下来来讨论在完整和不完整情况下的**执行流程**。

### 1.2.1 完整结构下，有2条执行主线

如果写的是如上个cell所示的完整结构，那么实际执行只有两条主线：

#### 1.2.1 (1) try没出错: `try → else → finally`

````python
try:
    print("Try branch is run.")
except:
    print("Except branch is run.")
else:
    print("Else branch is run.")
finally:
    print("Finally branch is run.")
````

#### 1.2.1 (2) try 出错，并且该错误能被 except 捕捉: `try → except → finally`

````python
try:
    print("Try branch is run.")
    with open("somefile.txt") as file:
        file.read()
except:
    print("This file doesn't exist. Except branch is run.")
else:
    file.write("Something")
    print("Else branch is run.")
finally:
    print("Finally branch is run.")
````

### 1.2.2 最小规则

当然，也可以不完整地写`try: except: else: finally:`，这种情况下，有3种写法需要关注：

#### 1.2.2 (1) `try...except`

```python
try:
    ...
except SomeKindOfError:
    ...
```
这种基本分成3种情况：

1. 如果 `try` 里面没有出错：

    - 执行 `try` 里面的代码，全部执行完
    - 跳过 `except`
    - 继续执行 `try-except` 结构后面的代码

2. 如果 `try` 里面出错了，而且被 `except` 捕捉到：

    - `try` 执行到出错的位置，**直接**跳到 `except`（后面的不处理了）
    - 执行 `except` 里面的代码
    - 然后继续执行 `try-except` 后面的代码

3. 如果 `try` 里面出错了，但没有被 `except` 捕捉到：

    - `try` 执行到出错的位置，**直接**报错，程序停止运行

#### 1.2.2 (2) `try...finally`

```python
try:
    ...
finally:
    ...
```
这种写法本身不处理错误。它的作用不是 “抓住错误” ，而是 “做收尾工作” 。

执行情况是：


1. try 没出错
    - try
    - finally
    - 程序继续往下走

````python
try:
    print("Try branch is run.")
finally:
    print("Finally branch is run.")
````

2. try 出错

````python
try:
    print("Try branch is run.")
    with open("AnoootherFile.txt") as file:
        file.read()
    print("After the error.")
finally:
    print("Finally branch is run.")

print("Now we are out.")
````

观察上面的代码、打印信息、报错信息，可以得知，在 `try...finally` 里，如果 `try` 中途出错：

1. `try` 会在出错那一行停住。
2. `try` 里面出错位置后面的代码不再执行。
3. 立刻跳去执行 `finally`。
4. `finally` 执行完之后，原来的错误继续抛出。
5. 如果没有别的 `except` 处理它，程序就报错停止。
6. 所以 `try...finally` 结构后面的代码也不会执行。

这个例子里：

```python
with open("AnoootherFile.txt") as file:
```

这里已经出错了，所以这句后面的：

```python
    file.read()
print("After the error.")
```

都不会执行。

然后执行：

```python
finally:
    print("Finally branch is run.")
```

执行完 `finally` 后，`FileNotFoundError` 继续报出来，所以最后面的：

```python
print("Now we are out.")
```

也不会执行。

也就是说：**`finally` 只保证自己会执行，但它不会让程序恢复正常。**


3. 再来看一个例子。这个例子体现了 `try ... finally` 常用于 **“收尾工作”**，比如关闭文件、释放资源、断开连接

```python
try:
    file = open("data.txt")
    # do something
finally:
    file.close()
```

实际上，打开文件

意思是 **就算中间出错了，也先把文件关掉**。

但它不会像 `except` 那样处理错误。要处理错误，必须写：`try ... except ...`

# 2. Catch the exception for `FileNotFoundError`

````python
try:
    with open("a_file.txt") as file:
        file.read()
except:
    print("There was an error")
````

不过，上面这种写法不能真正地解决错误。

那么怎么解决错误呢？文件不存在的话，那创建一个不就得了？所以我们可以在`except`分支中创建一个file。

````python
try:
    file = open("a_file.txt")   # PS: 这句话和下面这2行的效果是一致的
    
    # with open("a_file.txt") as file:
    #     file.read()

except:
    file = open("a_file.txt", "w")
    # 在writing mode下，open会尝试寻找这个file，如果找不到就创建
````

### 2.1 假设`"a_file.txt"`存在

*:请先解除掉上面的code cell中的注释并且执行这个cell。

现在`"a_file.txt"`已经存在了，如果这时候写下面这么一个代码块：

````python
try:
    file = open("a_file.txt")
    a_dictionary = {"key": "value"}
    print(a_dictionary["ksdjfsn"])

except:
    file = open("a_file.txt", "w")
    file.write("Something")
````

#### (1) 执行过程是：

```text
file = open("a_file.txt")        # 成功
a_dictionary = {"key": "value"}  # 成功
print(a_dictionary["ksdjfsn"])   # KeyError
```

然后因为写了 `except`，所以这个 `KeyError` 也会被抓住，程序会跳到 `except` 分支中去执行文件的打开和写入。

结果就是：**明明真正的问题是 dictionary 没有这个 key，但程序不会报 KeyError，而是去执行文件创建/写入逻辑。** 这就叫“KeyError 被忽略/吞掉了”。

#### (2) 问题核心

`except:`这个写法太宽了。它会把 `FileNotFoundError`、`KeyError` 等等都当成同一种错误处理。

更合理的是只捕捉真正想处理的错误：

````python
try:
    file = open("a_file.txt")
    a_dictionary = {"key": "value"}
    print(a_dictionary["ksdjfsn"])

except FileNotFoundError:
    file = open("a_file.txt", "w")
    file.write("Something")
````

这样如果文件不存在，才创建文件；如果是 `KeyError`，程序会正常报错，不会被误吞。

### 3. Catch the exception for `KeyError`

````python
try:
    file = open("a_file.txt")
    a_dictionary = {"key": "value"}
    print(a_dictionary["ksdjfsn"])

except FileNotFoundError:
    file = open("a_file.txt", "w")
    file.write("Something")

except KeyError as error_message:
    print(f"The key {error_message} does not exist.")
````

### 4. `else`

````python
try:
    file = open("a_file.txt")
    a_dictionary = {"key": "value"}
    print(a_dictionary["ksdjfsn"])

except FileNotFoundError:
    file = open("a_file.txt", "w")
    file.write("Something")

except KeyError as error_message:
    print(f"The key {error_message} does not exist.")

else:
    content = file.read()
    print(content)
````

### 5. `finally`

````python
try:
    file = open("a_file.txt")
    a_dictionary = {"key": "value"}
    print(a_dictionary["ksdjfsn"])

except FileNotFoundError:
    file = open("a_file.txt", "w")
    file.write("Something")

except KeyError as error_message:
    print(f"The key {error_message} does not exist.")

else:
    content = file.read()
    print(content)

finally:
    file.close()
    print("File was closed.")
````
