### 3.1  The constructor & the `next_question()` method
#### (1) The constructor
- Write the `__init__(...)` method.
- Initialize the `question_number` to `0`.
- Initialize the `question_list` to an input.

When creating this method, keep in mind that when we are calling the `QuizBrain(...)` constructor in the `main.py`, we are going to pass the `question_bank` into the `question_list` attribute.

```python
class QuizBrain1:

    def __init__(self, question_list):
        self.question_number = 0
        self.question_list = question_list
```

#### (2) The `next_question()` method

- Retrieve the item at the current `question_number` from the `question_list`.
- Use the `input()` function to show the user the Question text and ask for the user's answer.

```python
class QuizBrain2:

    def __init__(self, question_list):
        self.question_number = 0
        self.question_list = question_list

    def next_question(self):
        current_question = self.question_list[self.question_number]
        self.question_number += 1
        input(f"Q.{self.question_number}: {current_question.text} (True/False)?: ")
```

Now we can import the `QuizBrain` class into the `main.py` and create a `QuizBrain` class object to start the process of a quiz and ask a question.

```python
# in the main.py
from question_model import Question  # import the Question class
from data import question_data  # import the data to be processed
from quiz_brain import QuizBrain  # import the QuizBrain class


question_bank = []     # create an empty list to store Question objects

for dictionary in question_data:
    new_question = Question(dictionary["text"], dictionary["answer"])
    question_bank.append(new_question)

quiz = QuizBrain(question_bank)
quiz.next_question()
```

### 3.2 How to Continue Showing New Questions: `still_has_questions` method
> in the `quiz_brain.py` and `main.py`

#### (1) In the `quiz_brain.py`
- Create method called `still_has_questions()`.
- Return a boolean depending on the value of `question_number`.

```python
class QuizBrain3:

    def __init__(self, question_list):
        self.question_number = 0
        self.question_list = question_list

    def next_question(self):
        current_question = self.question_list[self.question_number]
        self.question_number += 1
        input(f"Q.{self.question_number}: {current_question.text} (True/False)?: ")

    def still_has_questions(self):
        return self.question_number < len(self.question_list)
```

#### (2) In the `main.py` 
- Use the `while` loop to show the next question until the end.


```python
# in the main.py
from question_model import Question  # import the Question class
from data import question_data  # import the data to be processed
from quiz_brain import QuizBrain  # import the QuizBrain class


question_bank = []     # create an empty list to store Question objects

for dictionary in question_data:
    new_question = Question(dictionary["text"], dictionary["answer"])
    question_bank.append(new_question)

quiz = QuizBrain(question_bank)
while quiz.still_has_questions():
    quiz.next_question()
```

### 3.3 Checking Answers and Keeping Score: `check_answers` method
> in the `quiz_brain.py`

#### (1) Check the answer
- Write a `check_answers` method and call it in the `next_question` method to see if the input answer is correct.

```python
class QuizBrain4:

    def __init__(self, question_list):
        self.question_number = 0
        self.question_list = question_list

    def next_question(self):
        current_question = self.question_list[self.question_number]
        self.question_number += 1
        user_answer = input(f"Q.{self.question_number}: {current_question.text} (True/False)?: ") # use variable `user_answer` to save the answer user input
        self.check_answer(user_answer, current_question.answer) # call this check_answer to check whether the answer is t/f

    def still_has_questions(self):
        return self.question_number < len(self.question_list)

    def check_answer(self, user_answer, correct_answer):
        if user_answer.lower() == correct_answer.lower():
            print("You got it right!")
        else:
            print("That's wrong.")
        print(f"The correct answer was: {correct_answer}.")
```

#### (2) Count the score
- Add a `score` attribute to the `QuizBrain` class.
- Increase the `score` attribute when user inputs correct answer.
- Print out the current `score` in the console for viewing.

```python
class QuizBrain5:

    def __init__(self, question_list):
        self.question_number = 0
        self.question_list = question_list
        self.score = 0

    def next_question(self):
        current_question = self.question_list[self.question_number]
        self.question_number += 1
        user_answer = input(f"Q.{self.question_number}: {current_question.text} (True/False)?: ") # use variable `user_answer` to save the answer user input
        self.check_answer(user_answer, current_question.answer) # call this check_answer to check whether the answer is t/f

    def still_has_questions(self):
        return self.question_number < len(self.question_list)

    def check_answer(self, user_answer, correct_answer):
        if user_answer.lower() == correct_answer.lower():
            self.score +=1
            print("You got it right!")
        else:
            print("That's wrong.")
        print(f"The correct answer was: {correct_answer}.")
        print(f"Your current score is: {self.score}/{self.question_number}\n\n")
```

### 3.4 The End of the Quiz

Print two lines to tell users that the quiz has ended and their final mark.

```python
class QuizBrain6:    # the final version

    def __init__(self, question_list):
        self.question_number = 0
        self.question_list = question_list
        self.score = 0

    def next_question(self):
        current_question = self.question_list[self.question_number]
        self.question_number += 1
        user_answer = input(f"Q.{self.question_number}: {current_question.text} (True/False)?: ") # use variable `user_answer` to save the answer user input
        self.check_answer(user_answer, current_question.answer) # call this check_answer to check whether the answer is t/f

    def still_has_questions(self):
        return self.question_number < len(self.question_list)

    def check_answer(self, user_answer, correct_answer):
        if user_answer.lower() == correct_answer.lower():
            self.score +=1
            print("You got it right!")
        else:
            print("That's wrong.")
        print(f"The correct answer was: {correct_answer}.")
        print(f"Your current score is: {self.score}/{self.question_number}\n\n")

    def show_the_result(self):
        print("You've completed the quiz.")
        print(f"You final score was: {self.score}/{self.question_number}")
```

And once again, call this function in the `main.py`.

```python
# <in the main.py>
quiz.show_the_result()
```
