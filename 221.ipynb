# 221. Working with Images and Setting up

这一课主要想讨论 window padding widgets 之间的距离关系和 window 的尺寸。

```python
from tkinter import *
YELLOW = "#f7f5dd"
```

```python
# Windows
window = Tk()
window.title("Password Manager")
```

```python
window.config(padx=20, pady=20, bg=YELLOW) # widgets 会被放在离 window 内边缘 20 像素之后的区域里
```

这种写法下，window 本身没有固定尺寸，而是因为 padding 会参与布局计算，让 window 自动变大来容纳 widget + padding。

```python
window.config(width=500, height=500, padx=20, pady=20, bg=YELLOW) 
```

但即使像这句代码里换用了另一个大小，最后打印出来的结果也是一样的。

```python
window.geometry("500x500")
window.config(padx=20, pady=20, bg=YELLOW)
```

## window的尺寸设置

### (1) `window.config`

window.config(width=..., height=...) 不是可靠地固定主窗口大小。

在这个例子里，窗口最终还是主要被 canvas.pack() 之后的 Canvas 尺寸撑开，所以两种显示结果一样。

### (2) `window.geometry`
