## 1. Create the User class

In Python, if you declare a class or a function with nothing in it, python won't like that, it would report an error "indent expected".

To tackle this problem, you can just write `pass` in the class or function.

````python
class User_1:
    pass

def function_1():
    pass
````

As we have created a `User_1` class with nothing in it, if we want to add attribute into it, we can use the ***dot notation method***:

Use a dot after the object and add in any piece of data we want.

````python
user_1 = User_1()
user_1.id = "001"
user_1.username = "Angela"

print(user_1.username)
````

In this way, we add attributes `id` and `username` to the **object** `user_1`.

However, it's quite obvious that this method is very inconvenient: For every object you need to define every attribute. That's troublesome and quite easy to make mistake.

### 1.1 Constructor
> A Constructor initializes an object.
> 
> Which means it sets variables, counters, switches, etc. to their starting values at the beginning of a program or subprogram.

In Python, the way that we would create the Constructor is by using a special function which is the `__init__` function:
```python
class Car:
    def __init__(self):
    # initialize attributes
```

IMPORTANT:
Everytime you create a new object, the `__init__` function would be called.

For example:

````python
class User_2:
    def __init__(self):
        print("new user being created...")

user_1 = User_2()
user_1.id = "001"
user_1.username = "Angela"

print(user_1.username)

user_2 = User_2()
````

You can see the above program printed two lines of "new user being created", which solidifies the fact that everytime the we create a new object, the `__init__` function is called and this time it prints a line into the console.

### 1.2 How do we set Attributes IN the Constructor
This is how our code would look like:
```python
class Car:
    def __init__(self, seats):
        self.seats = seats
```
First, the`self` is **the actual object that's being created**.

In addition to the `self`, you can add as many parameters as you wish (like the `seat` here) and that parameter is going to be passed in when an object gets constructed from this class. And once you received that data (in the parentheses right after the `__init__`), then you can use it to set the object's attribute.

This is what it would look like when we actually call our constructor:
```python
my_car = Car(5)
```
1. We know that we can create an object by calling the name of a class and then adding the parentheses.
2. But now inside our `__init__` function we have some additional parameters that we can also PASS IN data to those parameters which will be used to set the attributes for that object.

So in this case, once this line of code has completed, then `my_car.set` is gonna be equal to `5`.

This is exactly the same as this:
```python
my_car = Car()
my_car.seats = 5
```

But it does make it a lot quicker when you are creating a lot of objects that have the exactly the same attributes.

Now we can define the class:

````python
class User_3:
    
    def __init__(self, user_id, user_name):
        self.id = user_id
        self.user_name = user_name
````

By convention, you can set the name of the parameter (things in the parentheses) be equal to the name of the attribute (things inside the function body), but you don't really need to stick to this rule.

````python
user_1 = User_3("001", "Angela")
````

Remember, when you add parameters to the constructor, which is the `__init__` function, you are now saying that whenever a new object is being constructed from this class, it must provide these two pieces of data.

### 1.3 Provide a default value

It doesn't make sense for all attributes to be initialized when we actually create our object.

For example, if this `User` class is for some social media like instagram:
1. Every account's `followers` starts from `0`.
2. It would be stupid and wasteful to pass in the `0` for the parameter `followers` everytime we create a new object from `User` class.
Directly setting the attribute `followers`'s DEFAULT VALUE to `0` inside the body of the constructor is wise.

````python
class User_4:
    
    def __init__(self, user_id, user_name):
        self.id = user_id
        self.user_name = user_name
        self.followers = 0
````

Note that in this case we don't even need to declare the `followers` parameter in the parentheses of `__init__`, which is understandable because we don't need programmers to pass in the value of `followers` anymore.

````python
user_1 = User_4("001", "Angela")
print(user_1.followers)
````

Although we didn't pass in a value for attribute `followers`, it still exists and has its own value.

### 1.4 Add methods to a Class

The function `enter_race_mode` can modify the attribute `seats`.

What differs methods from functions is that methods ALWAYS take in a `self` as parameter, which indicates that the method is ASSOCIATED WITH the CURRENT OBJECT.

````python
class Car:
    def __init__(self, seats):
        self.seats = seats
        
    def enter_race_mode(self):
        self.seats = 2

my_car = Car(5)
my_car.enter_race_mode()
print(my_car.seats)
````

Now, we can develop the `User` class further. Say we want to add a follow function to this class.

We define a new method called `follow`. Besides `self`, this method also takes a `User` class **object** as the parameter, which is the `user` in the parentheses.

This `user` is the one we want to follow.

````python
class User:
    def __init__(self, user_id, username):
        self.id = user_id
        self.username = username
        self.followers = 0
        self.following = 0
    
    def follow(self, user):  # This `user` is a User class object
        user.followers +=1
        self.following +=1
````

As this `user` parameter is a `User` class object, we can visit its attribute `followers` by using the dot notatioin:
```python
user.followers +=1
```
Well, this user is being followed, which means its `followers` attribute will go up by one, which is quite straightforward.

And the `self`, which is the current object (the user who initiates the following action), as has followed `user`, its `following` attribute will also go up by one.

````python
user_1 = User("001", "Angela")
user_2 = User("002", "Jack")

user_1.follow(user_2)

print(user_1.followers)
print(user_1.following)
print(user_2.followers)
print(user_2.following)
````

line 4 in the above code:
1. Calls the `user_1`'s method `follow`, which means `user_1` initiates the calling act.
2. Passes in the object `user_2` to the `follow`'s parameter `user`.
3. Inside `follow` method, modifies `user_2`'s `followers` attribute and `user_1`'s `following` attribute.
