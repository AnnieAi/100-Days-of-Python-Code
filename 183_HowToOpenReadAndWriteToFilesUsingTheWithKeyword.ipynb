# 最原始的写法 & `read()` & `close()`

`open()`是 built-in 方法。

````python
file = open("my_file.txt")
contents = file.read()
print(contents)
file.close()
````

# `with...as`

下面这种写法，就不需要加`.close()`了。

在使用完文件后，会自动关闭。

````python
with open("my_file.txt") as file:
    contents = file.read()
    print(contents)
````

# `write()`
## 没有指定`mode`
下面这些么写会报错，因为文件是被以 "read-only" 状态打开的。

````python
with open("my_file.txt") as file:
    file.write("New data.")
````

这是因为，如果去看官方文档，我们会发现open的定义如下：
```python
open(file, mode='r')
```
意思就是说，默认情况下，`open`被设置为`'r'`模式打开。

（官方文档这里定义用的是单引号，双引号也可以。Python里面字符串既可以用单引号框起来，也可以用双引号框起来。

## 指定`mode="w"`

在`open()`里面添加`"w"`参数，意思是以“写入”模式打开。

这个状态下就可以写入了。

但是注意，这种情况下，写入的数据会【覆盖】原数据。

````python
with open("my_file.txt", "w") as file:
    file.write("New data.")
````

## 指定`mode="a"`

如果不想要原本的数据被覆盖掉，那就用`"a"`，a代表 "append"。就像之前给 list 用的`list.append`。

````python
with open("my_file.txt", "a") as file:
    file.write("\nNow there must be a line above.")
````

## 文件的创建
满足以下两个条件时，文件会被创建：
1) open里面的文件不存在
2) `mode="w"`（要是写入模式）

````python
with open("my_new_new_file.txt", "a") as file:
    file.write("A brand new file.")
````
